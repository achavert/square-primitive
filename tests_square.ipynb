```python
from square import *
import qibo
from qibo import Circuit, gates
```

```python
m = 3
n = 2*m
num_qubits = 2*n + 1

sq, _ = Square(m, method = 'ripple_input_carry', input_swaps = True)

circ = Circuit(num_qubits)
circ.add(sq.on_qubits(*range(num_qubits)))
circ.add(gates.M(*range(num_qubits)))

res = circ(nshots=1)
samples = res.samples()[0]
print(samples)
```

```python
from square import *

m = 5
n = 2 * m
num_qubits = 2 * n + 1

for x in range(2**m):
    sq, (R, F, Z) = Square(m, input_swaps = False)

    circ = initialize_square(x, m, allocs = (R, F, Z), direct=True)
    circ.add(sq.on_qubits(*range(num_qubits)))
    circ.add(gates.M(*range(num_qubits)))

    res = circ(nshots=1)

    samples = res.samples()[0]
    square_bits = samples[R[0]:R[n-1]+1]
    square_value = sum(bit << i for i, bit in enumerate(square_bits))
    print(f'Input: {x} | Result: {square_value} = {samples}')
```
