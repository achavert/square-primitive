```python
import qibo
from qibo import Circuit, gates, set_backend
import time
set_backend("qibojit", platform="cuda")
```

```python
from square_no_anc import *
m = 5
n = 2 * m
num_qubits = 2*n

start = time.time()
for x in range(2**m):
    sq, (R, F) = Square(m, input_swaps = False)

    circ = initialize_square(x, m, allocs=(R, F), direct=True)
    circ.add(sq.on_qubits(*range(num_qubits)))
    circ.add(gates.M(*range(num_qubits)))

    res = circ(nshots=1)

    samples = res.samples()[0]
    square_bits = samples[R[0]:R[n-1]+1]
    square_value = sum(bit << i for i, bit in enumerate(square_bits))
    print(f'Input: {x} | Result: {square_value} = {samples}')
end = time.time()
print(f'Elapsed time for {2**m} entries: {end-start:0.4f}')
```
