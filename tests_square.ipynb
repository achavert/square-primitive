```python
from square import *
import qibo
from qibo import Circuit, gates
```

```python
m = 3
n = 2*m
num_qubits = 2*n + 1

qft_circ = qibo.models.qft.QFT(m) 
qft_circ_inv = qibo.models.qft.QFT(n).invert()
sq = Square(m)

circ = Circuit(num_qubits)
circ.add(qft_circ.on_qubits(*range(m)))
circ.add(sq.on_qubits(*range(num_qubits)))
circ.add(qft_circ_inv.on_qubits(*range(n)))
circ.add(gates.M(*range(num_qubits)))
```

```python
res = circ(nshots=1)
samples = res.samples()[0]
print(samples)
```

```python
m = 3
n = 2*m
num_qubits = 2*n + 1

sq = Square(m, method = 'input_carry')

circ = Circuit(num_qubits)
#circ.add(gates.X(0))
circ.add(gates.X(1))
circ.add(gates.X(2))
circ.add(sq.on_qubits(*range(num_qubits)))
circ.add(gates.M(*range(num_qubits)))

res = circ(nshots=4)
samples = res.samples()
print(samples)
```

```python
from square import *
import qibo
from qibo import gates

m = 4
n = 2*m
num_qubits = 2*n + 1

for x in range(2**m):
    sq = Square(m)
    circ = initialize_square(x, m)
    circ.add(sq.on_qubits(*range(num_qubits)))
    circ.add(gates.M(*range(num_qubits)))
    res = circ(nshots=1)
    samples = res.samples()[0]
    R, F, Z = get_allocs(n)
    square_bits = samples[R[0]:R[n-1]+1]
    square_value = sum(bit << i for i, bit in enumerate(square_bits))
    print(f'Input: {x} | Result: {square_value} = {samples}')
```
