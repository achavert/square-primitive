```python
import qibo
import time
from square_np1 import *
qibo.set_backend("qibojit", platform="cuda")
```

```python
def run_circ(m):
    num_qubits = 2*m + 1
    start = time.time()
    for x in range(2**m):
        sq, (R, F) = Square(m, input_swaps = False)

        circ = initialize_square(x, m, final_allocs=(R, F), direct=True)
        circ.add(sq.on_qubits(*range(num_qubits)))
        circ.add(qibo.gates.M(*range(num_qubits)))

        res = circ(nshots=1)

        samples = res.samples()[0]
        square_bits = samples[R[0]:R[2*m-1]+1]
        square_value = sum(bit << i for i, bit in enumerate(square_bits))
        print(f'Input: {x} | Result: {square_value} = {samples} | res==x^2 {int(x**2)==int(square_value)}')
    end = time.time()
    print(f'Mean elapsed time: {(end-start)/2**m:0.4f} s')
```

```python
m_range = range(2,11)
for m in m_range:
    print('====================================')
    print(f'Input: m={m} bits')
    run_circ(m)
```
